```python
import pandas as pd

data = {
    "1": [3, 5, 12, 7, 1],
    "2": [11, 18, 22, 9, 4],
    "3": [18, 25, 30, 13, 8],
    "4": [22, 28, 33, 19, 10],
    "5": [31, 34, 35, 26, 21],
    "6": [37, 36, 36, 34, 25],
    "strong": [6, 2, 5, 4, 3]
}

df = pd.DataFrame(data)
df.to_csv("Lotto.csv", index=False)

df.head()
```

```python
df = pd.read_csv("Lotto.csv")

print("Shape:", df.shape)
print("\nColumns:", df.columns)

df.head()
```

## Step 4 — Data Cleaning

The dataset already contains valid Lotto numbers.
We only ensure all values are numeric and within valid ranges.

```python
nums = ["1","2","3","4","5","6"]

# convert to numeric (safety)
df[nums] = df[nums].apply(pd.to_numeric, errors="coerce")
df["strong"] = pd.to_numeric(df["strong"], errors="coerce")

# filter valid lotto ranges
df_clean = df[
    (df[nums].ge(1).all(axis=1)) &
    (df[nums].le(37).all(axis=1)) &
    (df["strong"].between(1,7))
]

df_clean.head()
```

## Step 5 — Frequency Analysis

We count how often each number appears in the dataset.
This helps identify frequently occurring ("hot") numbers.

```python
from collections import Counter

# collect all numbers
numbers = df_clean[nums].values.flatten()

# count frequency
main_counts = Counter(numbers)

# convert to dataframe
freq_df = pd.DataFrame(
    sorted(main_counts.items()),
    columns=["number","count"]
)

freq_df
```

```python
freq_sorted = freq_df.sort_values(by="count", ascending=False)

freq_sorted.head(5)
```

## Step 6 — Visualization

We visualize number frequencies to better understand the distribution.

The scoring system rewards:
- frequent numbers
- good sum range
- balanced odd/even ratio

It penalizes:
- consecutive numbers
- extreme sums

## Step 11 — Smart Strong Selection

We improve the selection of the strong number by using frequency-based weights.

```python
def choose_strong():
    strong_range = list(range(1,8))
    weights = [strong_counts.get(n, 0) + 1 for n in strong_range]

    return random.choices(strong_range, weights=weights, k=1)[0]
```

## Step 12 — Final AI Generator

This final generator creates many candidate lines, scores them, and returns the best-ranked lines.

The score considers:
- number frequency
- sum range
- odd/even balance
- consecutive number penalty
- strong number frequency

```python
def score_candidate(candidate, strong):
    score = 0

    # reward frequent numbers
    for n in candidate:
        score += main_counts.get(n, 0)

    # sum range
    total = sum(candidate)
    if 100 <= total <= 145:
        score += 5
    elif 90 <= total <= 160:
        score += 2
    else:
        score -= 10

    # odd/even balance
    odd = sum(n % 2 for n in candidate)
    if 2 <= odd <= 4:
        score += 3
    else:
        score -= 5

    # consecutive numbers penalty
    candidate = sorted(candidate)
    for i in range(len(candidate) - 1):
        if candidate[i+1] - candidate[i] == 1:
            score -= 3

    # strong number frequency reward
    score += strong_counts.get(strong, 0)

    return score


def generate_final_candidates(n_trials=5000, top_k=10):
    results = []

    for _ in range(n_trials):
        nums, strong = generate_candidate()
        total = sum(nums)
        score = score_candidate(nums, strong)

        results.append((nums, strong, total, score))

    # sort by score
    results = sorted(results, key=lambda x: x[3], reverse=True)

    # remove duplicate number sets
    unique = []
    seen = set()

    for nums, strong, total, score in results:
        key = tuple(nums)
        if key not in seen:
            seen.add(key)
            unique.append((nums, strong, total, score))

        if len(unique) == top_k:
            break

    return unique


final_lines = generate_final_candidates(n_trials=5000, top_k=10)

for i, (nums, strong, total, score) in enumerate(final_lines, start=1):
    print(f"Line {i}: {nums} | Strong: {strong} | Sum: {total} | Score: {score}")
```
